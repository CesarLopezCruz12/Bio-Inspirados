```python
#secondOrder1dCA.pynb
#Author: Jorge Luis Rosas Trigueros
#Visualizes the evolution of a 1d CA with 1st and 2nd order rules
#Last modification: 17nov23 20:23
# https://tinyurl.com/3kt2xpfj

import cv2
import numpy as np
from IPython import display as display
import ipywidgets as ipw
import PIL
from io import BytesIO
from random import random
```

```python
N=100

cells0=N*[0]
cells1=N*[0]
cells2=N*[0]

for i in range(N):
  cells1[i]= 0 if random()>0.5 else 1

# cells1[50]=0
# cells1[51]=0
# cells1[52]=0
# cells1[53]=1

rule=[0,0,0,1,0,0,1,0]
#rule=[0,0,1,1,1,1,1,0] #W62
rule=rule[::-1]

def apply_rule(left,center,right):
  global rule

  return rule[right+2*center+4*left]


def my_xor(a,b):
  if a==b:
    return 0
  return 1

def iteration():
  global cells0,cells1,cells2,N

  # for i in range(1,N-1):
  #   cells0[i]=apply_rule(cells1[i-1],cells1[i],cells1[i+1])
  for i in range(1,N-1):
    # cells0[i]=apply_rule(cells1[i-1],cells1[i],cells1[i+1])
    cells0[i]=my_xor(apply_rule(cells1[i-1],cells1[i],cells1[i+1]),cells2[i])
#    print(apply_rule(cells1[i-1],cells1[i],cells1[i+1]),cells2[i],cells0[i])
  # print("iteration")
  # print(cells2)
  # print(cells1)
  # print(cells0)
#  cells2[:]=cells1[:] #for 2nd order
  cells1[:]=cells0[:]




x0=0
y0=0
maxX=500
maxY=500
color=(255,255,255)
margin=1
stride = (maxX - 2*margin)/N


def graph_cells(img,cells):
  global x0,y0,maxX,maxY,margin,color,stride,N
  r=5
  for i in range(N):
    if cells[i]:
      start=(int(x0+stride*i+margin),int(y0+margin))
      end=(int(x0+stride*(i+1)-margin), int(y0+stride-margin))
      cv2.rectangle(img,start,end,color,-1)


wIm = ipw.Image()
display.display(wIm)

img = np.zeros((500, 500, 3), dtype="uint8")

graph_cells(img,cells1)
y0+=stride

while y0<=maxY:
  iteration()
  graph_cells(img,cells0)
  y0+=stride

pilIm = PIL.Image.fromarray(img, mode="RGB")
with BytesIO() as fOut:
    pilIm.save(fOut, format="png")
    byPng = fOut.getvalue()

# set the png bytes as the image value;
# this updates the image in the browser.
wIm.value=byPng
```
