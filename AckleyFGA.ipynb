```python
#Ackley Function
import math
import random
import time
import matplotlib.pyplot as plt
import numpy as np
from IPython import display as display
from functools import cmp_to_key
#Chromosomes are 10 bits long
L_chromosome=30
N_chains=2**L_chromosome
#Lower and upper limits of search space
a=-5
b=5
crossover_point=int(L_chromosome/2)
def random_chromosome2(L_chromosome):
    chromosome=[]
    for i in range(0,L_chromosome):
        if random.random()<0.5:
            chromosome.append(0)
        else:
            chromosome.append(1)
    return chromosome
    
#Number of chromosomes
N_chromosomes=30
#probability of mutation
prob_m=0.5
F0=[]
fitness_values=[]
for i in range(0,N_chromosomes):
    F0.append(random_chromosome2(L_chromosome))
    fitness_values.append(0)
    
#binary codification (two variables)
def decode_chromosome2(chromosome):
    L_gen=len(chromosome)//2
    N_chains=2**L_gen
    gen1=chromosome[:L_gen]
    gen2=chromosome[L_gen:]
    value1=0
    value2=0
    for p in range(L_gen):
        value1+=(2**p)*gen1[-1-p]
        value2+=(2**p)*gen2[-1-p]
    x=a+(b-a)*float(value1)/(N_chains-1)
    y=a+(b-a)*float(value2)/(N_chains-1)
    return x,y
    
def f(x,y):
    return (-20.0*math.exp(-0.2 * math.sqrt(0.5 * (pow(x,2) + pow(y,2))))-math.exp(0.5 * (math.cos(2 * math.pi * x)+math.cos(2 * math.pi * y))) + math.e + 20)

def evaluate_chromosomes():
    global F0
    for p in range(N_chromosomes):
        vx,vy=decode_chromosome2(F0[p])
        fitness_values[p]=f(vx,vy)
    
def compare_chromosomes(chromosome1,chromosome2):
    vc1x,vc1y=decode_chromosome2(chromosome1)
    vc2x,vc2y=decode_chromosome2(chromosome2)
    fvc1=f(vc1x,vc1y)
    fvc2=f(vc2x,vc2y)
    if fvc1 > fvc2:
        return 1
    elif fvc1 == fvc2:
        return 0
    else: #fvg1<fvg2
        return -1
suma=float(N_chromosomes*(N_chromosomes+1))/2.
Lwheel=N_chromosomes*10
def create_wheel():
    global F0,fitness_values
    maxv=max(fitness_values)
    acc=0
    for p in range(N_chromosomes):
        acc+=maxv-fitness_values[p]
    fraction=[]
    for p in range(N_chromosomes):
        fraction.append( float(maxv-fitness_values[p])/acc)
        if fraction[-1]<=1.0/Lwheel:
            fraction[-1]=1.0/Lwheel
    ## print fraction
    fraction[0]-=(sum(fraction)-1.0)/2
    fraction[1]-=(sum(fraction)-1.0)/2
    ## print fraction
    wheel=[]
    pc=0
    for f in fraction:
        Np=int(f*Lwheel)
        for i in range(Np):
            wheel.append(pc)
        pc+=1
    return wheel
F1=F0[:]

def nextgeneration(b):
    F0.sort(key=cmp_to_key(compare_chromosomes))
    print( "Best solution so far:")
    x,y=decode_chromosome2(F0[0])
    print( "f(",x,",",y,")= ", f(x,y))
    #elitism, the two best chromosomes go directly to the next generation
    F1[0]=F0[0]
    F1[1]=F0[1]
    for i in range(0,int((N_chromosomes-2)/2)):
        roulette=create_wheel()
        #print("r: ",roulette)
        #Two parents are selected
        p1=random.choice(roulette)
        p2=random.choice(roulette)
        #Two descendants are generated
        o1=F0[p1][0:crossover_point]
        o1.extend(F0[p2][crossover_point:L_chromosome])
        o2=F0[p2][0:crossover_point]
        o2.extend(F0[p1][crossover_point:L_chromosome])
        #Each descendant is mutated with probability prob_m
        if random.random() < prob_m:
            o1[int(round(random.random()*(L_chromosome-1)))]^=1
        if random.random() < prob_m:
            o2[int(round(random.random()*(L_chromosome-1)))]^=1
        #The descendants are added to F1
        F1[2+2*i]=o1
        F1[3+2*i]=o2
    #The generation replaces the old one
    F0[:]=F1[:]
F0.sort(key=cmp_to_key(compare_chromosomes))
evaluate_chromosomes()
# Función para ejecutar la evolución durante un número especificado de generaciones
def evolve_generations(num_generations):
    for _ in range(num_generations):
        nextgeneration(None) # Simulamos un clic en el botón
        
# Bucle interactivo para ejecutar la evolución
while True:
    user_input = input("Ingrese el número de generaciones a evolucionar (o 'exit' para salir): ")
    if user_input.lower() == "exit":
        break # Salir del programa si se ingresa "exit"
    try:
        num_generations = int(user_input)
        evolve_generations(num_generations)
    except ValueError:
        print("Entrada inválida. Por favor, ingrese un número o 'exit' para salir.")
# Si llega a este punto, el programa ha terminado
print("Programa finalizado.")
```
