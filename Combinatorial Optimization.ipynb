```python
import random

max_weight = 10  # Adjust for the specific problem
weights = [2, 2, 3, 4, 6, 7, 9]  # Adjust for the specific problem
values = [1, 2, 4, 5, 5, 6, 7]  # Adjust for the specific problem

# Step 1: Chromosome Representation (0s and 1s)
def initialize_population(population_size, chromosome_length):
    return [[random.randint(0, 1) for _ in range(chromosome_length)] for _ in range(population_size)]

# Step 2: Fitness Function
def fitness(chromosome, weights, values, max_weight):
    total_weight = sum(w * x for w, x in zip(weights, chromosome))
    if total_weight > max_weight:
        return 0  # Penalize solutions exceeding the weight limit
    return sum(v * x for v, x in zip(values, chromosome))

# Step 3: Initialization
population_size = 100
chromosome_length = len(weights)  # Adjust for the specific problem

population = initialize_population(population_size, chromosome_length)

# Step 4: Genetic Operators
def selection(population, fitness_func, num_parents):
    selected_parents = random.sample(population, num_parents)
    return selected_parents

def crossover(parent1, parent2):
    # One-point crossover
    crossover_point = random.randint(1, len(parent1) - 1)
    child1 = parent1[:crossover_point] + parent2[crossover_point:]
    child2 = parent2[:crossover_point] + parent1[crossover_point:]
    return child1, child2

def mutate(chromosome, mutation_rate):
    # Flip a random bit with a given probability
    for i in range(len(chromosome)):
        if random.random() < mutation_rate:
            chromosome[i] = 1 - chromosome[i]

# Main GA Loop
num_generations = 50
mutation_rate = 0.01
for generation in range(num_generations):
    parents = selection(population, fitness, num_parents=10)
    new_population = []
    while len(new_population) < population_size:
        parent1, parent2 = random.sample(parents, 2)
        child1, child2 = crossover(parent1, parent2)
        mutate(child1, mutation_rate)
        mutate(child2, mutation_rate)
        new_population.extend([child1, child2])
    population = new_population

# Find the best solution in the final population
best_solution = max(population, key=lambda chrom: fitness(chrom, weights, values, max_weight))
best_fitness = fitness(best_solution, weights, values, max_weight)

print("Best solution:", best_solution)
print("Total value:", best_fitness)
```
